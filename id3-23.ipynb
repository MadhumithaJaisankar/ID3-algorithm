```python
def classify(tree, instance):
    if not isinstance(tree,dict):
        return tree
    feature = list(tree.keys())[0]
    feature_value = instance[feature]
    if feature_value in tree[feature]:
        return classify(tree[feature][feature_value], instance)
    else:
        return 'no'
```
